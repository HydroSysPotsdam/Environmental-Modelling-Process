```python
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt
from safepython import EET, sampling, util
```

```python
def additive(x, y, z):
    return x + y + np.random.uniform(-0.1, 0.1)
```

```python
r = 100
M = 3
lower = np.array([-1, -1, -1])
upper = np.array([ 1,  1,  1])
labels = ["x", "y", "z"]

X = sampling.OAT_sampling(r, M, distr_fun=sp.stats.uniform, distr_par=np.transpose([lower, upper - lower]).tolist(), samp_strat="lhs", des_type="trajectory")
Y = np.array([additive(*xs) for xs in X])

mi, sigma, EE = EET.EET_indices(r, lower.tolist(), upper.tolist(), X, Y, design_type="trajectory")
mi    = mi.round(10)
sigma = sigma.round(10)
```

```python
dydx = np.zeros((r, 3))
x    = np.zeros((r, 3))
y    = np.zeros((r, 3))

for j in range(r):
    for i, label in enumerate(labels):
        k = j*4 + i

        x[j, i] = X[k, i]
        y[j, i] = Y[k]

        dx = (X[k + 1, i] - X[k, i])
        dy = (Y[k + 1] - Y[k])
        
        dydx[j, i] = dy/dx
```

```python
EET.EET_plot(dydx.mean(axis=0).round(10), dydx.std(axis=0).round(10), labels)

fig, axs = plt.subplots(3, 3, figsize=(10, 8))

for i, label in enumerate(labels):
    color = f"C{i}"
    
    plt.sca(axs[0, i])
    #plt.scatter(X[:, i], Y, alpha=0.2)
    plt.scatter(x[:, i], y[:, i], color=color, alpha=0.2)
    plt.ylabel("Y")
    plt.xlabel(label)

    plt.sca(axs[1, i])
    plt.scatter(x[:, i], dydx[:, i], color=color, alpha=0.2)
    plt.xlabel(label)
    plt.ylabel("dy/dx")

    plt.sca(axs[2, i])
    plt.hist(dydx[:, i], bins=np.linspace(dydx.min(), dydx.max(), 50, endpoint=True), color=color)
    plt.title("EEs " + label)
    plt.ylabel("EE")
    
plt.tight_layout()
```
